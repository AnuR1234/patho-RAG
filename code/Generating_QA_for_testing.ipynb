```python
# Generate Random qnuestion based on Promt using openAI from the pdf text books and

import json
import random
import openai
import os
all_generated_questions = []
# Set your OpenAI GPT-3 API key
import os
from dotenv import load_dotenv
load_dotenv()
openai.api_key = os.getenv("OPENAI_API_KEY")
if not openai.api_key:
    raise RuntimeError("OPENAI_API_KEY is missing from .env")
output_file_path = '/Users/anureddy/Desktop/SEM03/Pratikums/patho-RAG/code/generated_questions_pdfs.json' # change the path accordingly 
def generate_questions_pdf(context, book_title, page_number, amount):

  '''We ask ChatGPT to form questions over our context, while taking into account different forms of questions.
       1. Confirmation Question [yes or no]
       2. Factoid-type Question [what, which, when, who, how]
       3. List-type Question
       4. Casual Question [why or how]
       6. Complex question (requires understanding of multiple texts)
  '''
  question_type = ["confirmation questions [yes or no]", "factoid-type questions [what, which, when, who, how]", "list-type questions", "casual questions [why or how]", "questions"]
    #question_type = ["confirmation questions [yes or no]", "factoid-type questions [what, which, when, who, how]", "list-type questions", "casual questions [why or how]", "hypothetical questions [e.g. what would happen if...]", "questions"]
  random_type = random.randint(0, len(question_type)-1)

  prompt = f"""Context: {context}\nBook Title: {book_title}\nPage Number: {page_number}\n\nForm {question_type[random_type]} in your own words. Answer in this json format {{"question_1": '', ...}} (Exact Amount: {amount}), don't say anything else."""

  response = openai.ChatCompletion.create(
        model="gpt-3.5-turbo",
        messages=[
            {"role": "system", "content": "assistant"},
            {"role": "user", "content": prompt},
        ],
        temperature=0.7,
        max_tokens=100,
        n=2
    )

  try:
        answer = json.loads(response['choices'][0]['message']['content'])
  except json.JSONDecodeError as e:
        print(f"Error with {prompt}\n{response}\n")
        return None
  return answer

def get_random_documents(file_path, num_documents=50):
    try:
        with open(file_path, 'r') as file:
            documents = json.load(file)  # Load the entire JSON array

        selected_documents = random.sample(documents, min(num_documents, len(documents)))
        return selected_documents
    except json.JSONDecodeError as e:
        print(f"Error parsing JSON file: {e}")
        return []
    except Exception as e:
        print(f"An error occurred: {e}")
        return []

# Modify the file path to your jsonl file
jsonl_file_path = '/Users/anureddy/Desktop/SEM03/Pratikums/pathlogy_pratikum/data/datasets_ids.json' # change the path
rand_docs = get_random_documents(jsonl_file_path, num_documents=10)

for document in rand_docs:
    context = document.get("page_content", "")
    book_title = document['metadata'].get("book_title", "")
    page_number = document['metadata'].get("page_number", "")

    generated_questions = generate_questions_pdf(context, book_title, page_number, amount=3)
    if generated_questions is not None:
        all_generated_questions.append({
            "context": context,
            "book_title": book_title,
            "page_number": page_number,
            "questions": generated_questions
        })

# Dump all generated questions to the output file
with open(output_file_path, 'w') as outfile:
    json.dump(all_generated_questions, outfile, indent=2)
```
